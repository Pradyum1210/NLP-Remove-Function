**Remove HTML Tags**

```python
import re

def remove_html_tags(text):
    pattern = re.compile('<.*?>')
    return pattern.sub('', text)

sample_html = "<h1>Hello</h1> This is <b>bold</b> text."
print(remove_html_tags(sample_html))
```

**Remove URLs / Links**

```python
import re

def remove_urls(text):
    pattern = re.compile(r'https?://\S+|www\.\S+')
    return pattern.sub('', text)


sample_url = "Visit my site https://google.com and http://example.com"
print(remove_urls(sample_url))
```

**Remove Emojis**

```python
import re

def remove_emojis(text):
    emoji_pattern = re.compile(
        "["
        "\U0001F600-\U0001F64F"  # emoticons
        "\U0001F300-\U0001F5FF"  # symbols & pictographs
        "\U0001F680-\U0001F6FF"  # transport & map
        "\U0001F1E0-\U0001F1FF"  # flags
        "\U00002702-\U000027B0"
        "\U000024C2-\U0001F251"
        "]+",
        flags=re.UNICODE
    )
    return emoji_pattern.sub('', text)



sample_emoji = "I love Python 😍🔥👍"
print(remove_emojis(sample_emoji))
```

**Remove Special Characters**

```python
import re

def remove_special_char(text):
    pattern = re.compile(r'[^a-zA-Z0-9\s]')
    return pattern.sub('', text)


sample_special = "Hello!!! This @@ is ## Python $$"
print(remove_special_char(sample_special))
```

**Remove Numbers**

```python
import re

def remove_numbers(text):
    pattern = re.compile(r'\d+')
    return pattern.sub('', text)


sample_numbers = "My roll number is 12345 and age is 20."
print(remove_numbers(sample_numbers))
```

**Remove Extra Spaces**

```python
import re

def remove_extra_spaces(text):
    return re.sub(r'\s+', ' ', text).strip()


sample_spaces = "This    sentence     has   too   many spaces."
print(remove_extra_spaces(sample_spaces))
```

**Convert to Lowercase**

```python
def to_lower(text):
    return text.lower()


sample_lower = "HELLO PYTHON WORLD"
print(to_lower(sample_lower))
```

**Remove Stopwords**

```python
import nltk
from nltk.corpus import stopwords
nltk.download('stopwords')

def remove_stopwords(text):
    stop_words = set(stopwords.words('english'))
    return " ".join([word for word in text.split() if word.lower() not in stop_words])


sample_stop = "This is the best example and it is easy to understand."
print(remove_stopwords(sample_stop))
```

**Tokenize Text**

```python
import re

def tokenize(text):
    return re.findall(r'\b\w+\b', text)


sample_token = "NLP is awesome! Let's learn tokenization."
print(tokenize(sample_token))
```

**Remove Punctuation**

```python
import re

def remove_punctuation(text):
    return re.sub(r'[^\w\s]', '', text)


sample_punc = "Hello, how are you? I'm fine!"
print(remove_punctuation(sample_punc))
```
